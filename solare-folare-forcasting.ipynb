```python
import os
import pickle
import numpy as np
import pandas as pd

pkl_path = "/kaggle/input/datasets/esayasmelaku/machine-learning-flare-dataset/ARTop_Processed_Regions (1).pkl"

print("File exists:", os.path.exists(pkl_path))

with open(pkl_path, "rb") as f:
    data = pickle.load(f)

print("Dataset type:", type(data))
print("Active regions:", len(data))
```

```python
ar_id = list(data.keys())[0]

print("Example AR ID:", ar_id)
print("Available components:", data[ar_id].keys())

series_df = data[ar_id]["series"]

print("\nSeries shape:", series_df.shape)
print("\nColumns:")
print(series_df.columns.tolist())

display(series_df.head())
```

```python
observation_counts = {
    ar_id: len(ar_data["series"])
    for ar_id, ar_data in data.items()
}

total_rows = sum(observation_counts.values())

print("Active regions:", len(data))
print("Total time-series observations:", total_rows)

print("\nObservation statistics:")
print(pd.Series(observation_counts).describe())
```

```python
column_signatures = {
    tuple(ar_data["series"].columns)
    for ar_data in data.values()
}

print("Number of different column structures:", len(column_signatures))

if len(column_signatures) == 1:
    print("All Active Regions have the same columns.")
else:
    print("WARNING: Different column structures exist.")
```

```python
quality_rows = []

for ar_id, ar_data in data.items():
    df = ar_data["series"]

    numeric_df = df.select_dtypes(include=np.number)

    quality_rows.append({
        "ar_id": ar_id,
        "n_obs": len(df),
        "n_missing": df.isna().sum().sum(),
        "n_inf": np.isinf(numeric_df).sum().sum(),
        "n_duplicate_rows": df.duplicated().sum()
    })

quality_df = pd.DataFrame(quality_rows)

print("Total missing values:", quality_df["n_missing"].sum())
print("Total infinite values:", quality_df["n_inf"].sum())
print("Total duplicate rows:", quality_df["n_duplicate_rows"].sum())

print("\nARs containing missing values:")
display(
    quality_df[quality_df["n_missing"] > 0]
    .sort_values("n_missing", ascending=False)
)
```

```python
suspect_ars = quality_df.loc[
    quality_df["n_missing"] > 0, "ar_id"
].tolist()

for ar_id in suspect_ars:
    df = data[ar_id]["series"]

    missing_by_column = df.isna().sum()
    missing_by_column = missing_by_column[missing_by_column > 0]

    print(f"\nAR {ar_id} | observations: {len(df)}")
    print(missing_by_column)
```

```python
feature_cols = [
    "bz",
    "vz",
    "sz",
    "dL",
    "dH",
    "vz_dL_pos",
    "vz_dH_pos",
    "vzbz"
]

target_cols = [
    "flare6hr",
    "flare12hr",
    "flare24hr",
    "flare48hr"
]

print("Physical features:")
for col in feature_cols:
    print(" -", col)

print("\nFuture flare targets:")
for col in target_cols:
    print(" -", col)
```

```python
feature_stats = []

for ar_id, ar_data in data.items():
    df = ar_data["series"]
    
    feature_stats.append(df[feature_cols])

all_features = pd.concat(feature_stats, ignore_index=True)

print("Total observations:", len(all_features))
print("Number of features:", len(feature_cols))

print("\nMissing values:")
display(all_features[feature_cols].isna().sum())

print("\nFeature statistics:")
display(all_features[feature_cols].describe().T)
```

```python
for target in target_cols:
    values = []

    for ar_data in data.values():
        values.extend(
            ar_data["series"][target].dropna().tolist()
        )

    s = pd.Series(values)

    print("\n" + "=" * 60)
    print(target)
    print("=" * 60)

    print("Count:", len(s))
    print("Minimum:", s.min())
    print("Maximum:", s.max())
    print("Mean:", s.mean())
    print("Median:", s.median())
    print("Unique values:", s.nunique())

    print("\nMost common values:")
    display(s.value_counts().head(20))
```

```python
flare24_values = []

for ar_data in data.values():
    flare24_values.extend(
        ar_data["series"]["flare24hr"].dropna().tolist()
    )

flare24 = pd.Series(flare24_values)

print("flare24hr observations:", len(flare24))
print("Minimum:", flare24.min())
print("Maximum:", flare24.max())
print("Mean:", flare24.mean())
print("Median:", flare24.median())
print("Unique values:", flare24.nunique())

print("\nTop 30 most common values:")
display(flare24.value_counts().head(30))
```

```python
tables = []

for ar_id, ar_data in data.items():
    df = ar_data["series"].copy()

    df["ar_id"] = ar_id

    tables.append(df)

combined_df = pd.concat(tables, ignore_index=True)

print("Combined dataset shape:", combined_df.shape)

display(combined_df.head())
```

```python
print("Rows:", len(combined_df))
print("Columns:", len(combined_df.columns))

print("\nColumns:")
print(combined_df.columns.tolist())

print("\nMissing values:")
display(combined_df.isna().sum())

print("\nDuplicate rows:")
print(combined_df.duplicated().sum())
```
